```python
# class , object , attribute , methods
class Car:
    wheels = 4     # Class attribute (shared by all instances)

    def __init__(self, brand, model):
        self.brand = brand   # instance attribute
        self.model = model

    def display(self):  # instance method
        return f"{self.brand}, {self.model}"


# creating objects
car1 = Car("BMW", "Top_model")
car2 = Car("Honda", "Civic")

print(car1.display())  # BMW, Top_model
print(car1.wheels)     # 4
```

```python
#Constructor: __init__() and self
class student:
    def __init__(self,name,age,grade):   # 'self' refers to the instance
        self.name = name
        self.age = age
        self.grade = grade

    def get_info(self):
        return (f"{self.name}, Age: {self.age}, grade: {self.grade}")
    
student = student("Alice", 20, "A")
print(student.get_info())  # Alice, Age: 20, Grade: A
    
```

```python
#inheritance
    #single inheritance
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return(f"{self.name} makes a sound")
    
class Dog(Animal):
    def speak(self):
        return(f"{self.name} barks")
dog = Dog("Buddy")
print(dog.speak())

```

```python
#multiple inheritence
class material:
    def __init__(self, name):
        self.name = name


class property(material):
    def properties(self, properties):
        return f"property of {self.name} is {properties}"


class steel(property):   # now inherits property → which already inherits material
    def use(self, use):
        return use


# Example usage
s = steel("Steel")

print(s.name)                    # Steel
print(s.properties("Hard"))      # property of Steel is Hard
print(s.use("Construction"))     # Construction
```

```python
#multilevel inheritence
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)  # Call parent constructor
        self.model = model

class ElectricCar(Car):
    def __init__(self, brand, model, battery_capacity):
        super().__init__(brand, model)
        self.battery_capacity = battery_capacity

tesla = ElectricCar("Tesla", "Model 3", "75 kWh")
print(tesla.brand)  # Tesla
```

```python
#super function
class Parent:
    def __init__(self, name):
        self.name = name
    
    def show(self):
        return f"Parent: {self.name}"

class Child(Parent):
    def __init__(self, name, age):
        super().__init__(name)  # Call parent's __init__
        self.age = age
    
    def show(self):
        # Call parent's show method
        parent_info = super().show()
        return f"{parent_info}, Age: {self.age}"

child = Child("John", 10)
print(child.show())  # Parent: John, Age: 10
```

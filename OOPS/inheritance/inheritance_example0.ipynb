```python
class Animal:
    def __init__(self, name, age, habitat):
        self.name, self.age, self.habitat = name, age, habitat
        self.energy = 100

    def eat(self): self.energy += 10
    def sleep(self): self.energy = 100

    def describe(self):
        print(f"\nName: {self.name}, Age: {self.age}, Habitat: {self.habitat}, Energy: {self.energy}")

    def speak(self):
        print(f"{self.name} makes a sound")


class Dog(Animal):
    def __init__(self, name, age, habitat, breed):
        super().__init__(name, age, habitat)
        self.breed = breed

    def speak(self): print(f"{self.name}: Woof")
    def fetch(self): self.energy -= 15

    def describe(self):
        super().describe()
        print(f"Breed: {self.breed}")


class Cat(Animal):
    def __init__(self, name, age, habitat, color):
        super().__init__(name, age, habitat)
        self.color = color

    def speak(self): print(f"{self.name}: Meow")
    def climb(self): self.energy -= 10

    def describe(self):
        super().describe()
        print(f"Color: {self.color}")


class Bird(Animal):
    def __init__(self, name, age, habitat, wing_span):
        super().__init__(name, age, habitat)
        self.wing_span = wing_span

    def speak(self): print(f"{self.name}: Tweet")
    def fly(self): self.energy -= 20

    def describe(self):
        super().describe()
        print(f"Wing Span: {self.wing_span}")


class Puppy(Dog):
    def __init__(self, name, age, habitat, breed, teeth):
        super().__init__(name, age, habitat, breed)
        self.teeth = teeth

    def speak(self): print(f"{self.name}: Yip")
    def play(self): self.energy -= 5

    def describe(self):
        super().describe()
        print(f"Teeth: {self.teeth}")


class TeethingPuppy(Puppy):
    def __init__(self, name, age, habitat, breed, teeth, toys):
        super().__init__(name, age, habitat, breed, teeth)
        self.toys = toys

    def speak(self): print(f"{self.name}: Grrr")
    def chew(self): self.energy -= 10

    def describe(self):
        super().describe()
        print(f"Toys: {self.toys}")


class Fish(Animal):
    def __init__(self, name, age, habitat, water):
        super().__init__(name, age, habitat)
        self.water = water

    def speak(self): print(f"{self.name}: Blub")
    def swim(self): self.energy -= 5

    def describe(self):
        super().describe()
        print(f"Water: {self.water}")


# ===== Testing =====
animals = [
    Dog("Buddy", 3, "House", "Labrador"),
    Cat("Kitty", 2, "House", "White"),
    Bird("Tweety", 1, "Forest", 20),
    Puppy("Max", 1, "House", "GS", 28),
    TeethingPuppy("Charlie", 0.5, "House", "Beagle", 20, 5),
    Fish("Nemo", 1, "Aquarium", "Salt")
]

for a in animals:
    a.describe()
    a.speak()
```

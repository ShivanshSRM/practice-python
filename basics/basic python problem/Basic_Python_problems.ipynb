```python
"""
Practice Problem: Write a Python function that accepts
two integer numbers. If the product of the two numbers
is less than or equal to 1000, return their product;
otherwise, return their sum.
"""
class solution():
    def accept(self, a, b):
        c = a * b
        if c <= 1000:
            return c
        else:
            return a + b
```

```python
"""
Practice Problem: Iterate through the first `num` numbers (0..num-1).
In each iteration, print the current number, the previous number, and their sum.
"""
def first10(self, num=10):
    prev = 0
    for i in range(num):
        s = i + prev
        print("Current:", i, "Previous:", prev, "Sum:", s)
        prev = i

# attach as a method in a later cell
solution.first10 = first10
```

```python
"""
Display only those characters which are present
at an even index number in given string.
"""
def slicing_solution3(self, s):
    for i in range(len(s)):
        if i % 2 == 0:
            print(s[i], i)

solution.slicing_solution3 = slicing_solution3

def slicing_solution4(self,s):
  print(s[0::2])
solution.slicing_solution4 = slicing_solution3
```

```python
"""
Write a function to remove characters
from a string starting from index 0 up
to n and return a new string.
"""
def remove_char(self,s,num):
  print(s[num:])
solution.remove_char = remove_char
```

```python
"""
Variable Swapping (The In-Place Method)-Write a
program to swap the values of two
variables, a and b, without using a
third temporary variable.
"""
def swapping(self,a,b):
  a,b= b,a
  print(a,b)
solution.swapping = swapping
```

```python
"""
Calculating Factorial with a Loop
Practice Problem: Write a program that
calculates the factorial of a given number
(e.g., 5!) using a for loop.
"""
def fact5(self,num):
  fact = 1
  for i in range(1,num+1):
      fact = fact*i
  print(fact)
solution.fact5 = fact5
```

```python
"""
List Manipulation: Add and Remove
Practice Problem: Create a list of 5 fruits.
Add a new fruit to the end of the list, then
remove the second fruit (at index 1).
"""
def soln7_list(self, list, fruit):
    list.append(fruit)     # add fruit to end
    list.pop(1)            # remove element at index 1
    print(list)

solution.soln7_list = soln7_list
```

```python
def reversekardo(self,s):
  print(s[::-1])
solution.reversekardo =reversekardo
```

```python
"""
Exercise 9. Vowel Frequency Counter
Practice Problem: Write a program to
count the total number of vowels (a, e,
i, o, u) present in a given sentence.
"""

def vowelfinder(self, s):
    count = 0
    listt= []
    for i in s:   # FIX: range(s) is wrong
        if i in ("a", "e", "i", "o", "u"):
            listt.append(i)
            count += 1
    print(count)
    print(listt)

solution.vowelfinder = vowelfinder
```

```python
"""
Exercise 10. Finding Extremes (Min/Max) in a List

Practice Problem: Given a list of integers, find and
print both the largest and the smallest numbers.
"""

# Fix for min_max shadowing built-in functions
def fixed_min_max(self, L):
  max_val = max(L)
  min_val = min(L)
  print(max_val, min_val)

solution.min_max = fixed_min_max
```

```python
"""
Exercise 11. Removing Duplicates from a List
Practice Problem: Write a script that takes a
list containing duplicate items and returns a
new list with only unique elements.
"""
def remove_duplicate(self, L):
  L = set(L)
  print(L)
solution.remove_duplicae = remove_duplicate
```

```python
"""
Exercise 12. List Comparison and Boolean Logic
Practice Problem: Write a function to return True
if the first and last number of a given list is
the same. If the numbers are different,
return False.
"""
def boleann(self, L):
  if L[0]== L[-1]:
    print(True)
  else:
    print("False")

solution.boleann = boleann
```

```python
"""
Exercise 13. Filtering Lists with
Conditional Logic Practice Problem:
Iterate through a given list of numbers
and print only those numbers which are
divisible by 5.
"""
def filterin_List(self,L):
  i = 0
  while i < len(L):
    if L[i] % 5 == 0:
      print(L[i])
    i +=1
solution.filterin_List = filterin_List
```

```python
"""
Exercise 14. Substring Frequency Analysis
Practice Problem: Write a program to find
how many times the substring “Emma” appears
in a given string.
"""
def Emma_finder(self, s):
  count = s.count("Emma")
  print(f"emma appeared  {count} times")
solution.Emma_finder = Emma_finder
```

```python
def pattern(self,num):
  for i in range(1,num+1):
    for j in range(i):
      print(i, end = " ")
    print()
solution.pattern = pattern
```

```python
def palindrome_check(self,num1):
  num1= str(num1)
  if num1 == num1[::-1]:
    print("Palindrome")
  else:
    print("no palindrome")

solution.palindrome_check= palindrome_check
```

```python
"""
Exercise 17. Merging Lists with
Parity Filtering . Practice Problem:
Create a new list from two given lists
such that the new list contains odd numbers
from the first list and even numbers from the s
econd list.
"""
def merge_list(self,l1,l2):
  l3 = []
  for i in l1:
    if i % 2 != 0:
      l3.append(i)

  for i in l2:
    if i % 2 == 0:
      l3.append(i)
  return l3

solution.merge_list = merge_list
```

```python
"""
Exercise 18. Integer Digit Extraction and Reversal
Practice Problem: Write a program to extract each
digit from an integer in the reverse order.
"""
def extract_dig(self, num):
  num = str(num)
  for i in range (len(num)):
    print(num[i])
solution.extract_dig = extract_dig
```
